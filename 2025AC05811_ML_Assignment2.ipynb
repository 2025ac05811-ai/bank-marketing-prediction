# Step 2: Machine Learning Classification Models and Evaluation

## 2.1 Data Preprocessing

Before training the Machine Learning classification models, the dataset is preprocessed to ensure it is suitable for analysis. The preprocessing activities include understanding the dataset, checking for missing values, encoding categorical variables, splitting the dataset into training and testing sets, and applying feature scaling where required.

These preprocessing steps prepare the dataset for building and evaluating the required Machine Learning models.

```python
# =============================================================================
# 2.1.1 Statistical Summary of Numerical Features
# =============================================================================

# Display summary statistics for all numerical columns
df.describe()
```

```python
# =============================================================================
# 2.1.2 Check for Missing Values
# =============================================================================

# Display the number of missing values in each column
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
# =============================================================================
# 2.1.3 Analyse the Target Variable
# =============================================================================

# Display the count of each class in the target variable
print("Target Variable Distribution:\n")
print(df['y'].value_counts())
```

```python
# =============================================================================
# 2.1.4 Encode Categorical Variables
# =============================================================================

from sklearn.preprocessing import LabelEncoder

# Create a LabelEncoder object
label_encoder = LabelEncoder()

# Identify all categorical columns
categorical_columns = df.select_dtypes(include='object').columns

# Apply Label Encoding to each categorical column
for column in categorical_columns:
    df[column] = label_encoder.fit_transform(df[column])

# Display the first five records after encoding
df.head()
```

```python
# =============================================================================
# 2.1.5 Split the Dataset into Features and Target Variable
# =============================================================================

# Separate the input features (X) and target variable (y)
X = df.drop('y', axis=1)
y = df['y']
print("Training Feature Order:")
print(X.columns.tolist())

# Display the shape of the feature matrix and target vector
print("Feature Matrix (X) Shape :", X.shape)
print("Target Variable (y) Shape:", y.shape)
```

```python
# =============================================================================
# 2.1.6 Split the Dataset into Training and Testing Sets
# =============================================================================

from sklearn.model_selection import train_test_split

# Split the dataset into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the shape of the training and testing datasets
print("Training Feature Set :", X_train.shape)
print("Testing Feature Set  :", X_test.shape)
print("Training Target Set  :", y_train.shape)
print("Testing Target Set   :", y_test.shape)
```

```python
# =============================================================================
# 2.1.7 Feature Scaling
# =============================================================================

from sklearn.preprocessing import StandardScaler

# Create a StandardScaler object
scaler = StandardScaler()

# Fit the scaler on the training data and transform it
X_train_scaled = scaler.fit_transform(X_train)

# Use the same scaler to transform the testing data
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully.")
```

## 2.2 Logistic Regression

Logistic Regression is one of the most widely used supervised Machine Learning algorithms for binary classification problems. It learns the relationship between the input features and the target variable and predicts the probability that a customer belongs to a particular class.

In this assignment, Logistic Regression is used to predict whether a customer will subscribe to a term deposit based on the customer attributes and previous marketing campaign information.

```python
# =============================================================================
# 2.2.1 Train the Logistic Regression Model
# =============================================================================

from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model
logistic_model = LogisticRegression(random_state=42)

# Train the model using the training dataset
logistic_model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully.")
```

```python
# =============================================================================
# 2.2.2 Predict the Test Dataset
# =============================================================================

# Predict the target values for the testing dataset
y_pred = logistic_model.predict(X_test_scaled)

print("Prediction completed successfully.")
```

```python
# =============================================================================
# 2.2.3 Evaluate Logistic Regression - Accuracy
# =============================================================================

from sklearn.metrics import accuracy_score

# Calculate the accuracy of the Logistic Regression model
logistic_accuracy = accuracy_score(y_test, y_pred)

print(f"Logistic Regression Accuracy: {logistic_accuracy:.4f}")
```

```python
# =============================================================================
# 2.2.4 Evaluate Logistic Regression - Precision, Recall and F1 Score
# =============================================================================

from sklearn.metrics import precision_score, recall_score, f1_score

# Calculate evaluation metrics
logistic_precision = precision_score(y_test, y_pred)
logistic_recall = recall_score(y_test, y_pred)
logistic_f1 = f1_score(y_test, y_pred)

# Display the evaluation metrics
print(f"Precision : {logistic_precision:.4f}")
print(f"Recall    : {logistic_recall:.4f}")
print(f"F1 Score  : {logistic_f1:.4f}")
```

```python
# =============================================================================
# 2.2.5 Evaluate Logistic Regression - AUC Score and MCC
# =============================================================================

from sklearn.metrics import roc_auc_score, matthews_corrcoef

# Calculate prediction probabilities for the positive class
y_prob = logistic_model.predict_proba(X_test_scaled)[:, 1]

# Calculate AUC Score
logistic_auc = roc_auc_score(y_test, y_prob)

# Calculate Matthews Correlation Coefficient (MCC)
logistic_mcc = matthews_corrcoef(y_test, y_pred)

# Display the evaluation metrics
print(f"AUC Score : {logistic_auc:.4f}")
print(f"MCC Score : {logistic_mcc:.4f}")
```

The Logistic Regression model achieved a good overall accuracy (88.79%). However, the recall was relatively low (21.72%), indicating that the model identified only a small proportion of customers who actually subscribed to the term deposit. This suggests that although the model performed well overall, it was less effective in detecting the minority class.

## 2.3 Decision Tree Classifier

Decision Tree is a supervised Machine Learning algorithm used for both classification and regression problems. It makes predictions by recursively splitting the dataset into smaller subsets based on the most informative features, forming a tree-like structure of decision rules. Unlike Logistic Regression, Decision Tree does not require feature scaling because it makes decisions based on feature values rather than distances.

In this assignment, the Decision Tree classifier is trained to predict whether a customer will subscribe to a term deposit based on customer information and previous marketing campaign details.

```python
# =============================================================================
# 2.3.1 Train the Decision Tree Classifier
# =============================================================================

from sklearn.tree import DecisionTreeClassifier

# Create the Decision Tree model
decision_tree_model = DecisionTreeClassifier(random_state=42)

# Train the model using the training dataset
decision_tree_model.fit(X_train, y_train)

print("Decision Tree model trained successfully.")
```

```python
# =============================================================================
# 2.3.2 Predict and Evaluate the Decision Tree Classifier
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    matthews_corrcoef
)

# Predict the target values for the testing dataset
dt_y_pred = decision_tree_model.predict(X_test)

# Predict probabilities for AUC calculation
dt_y_prob = decision_tree_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
dt_accuracy = accuracy_score(y_test, dt_y_pred)
dt_precision = precision_score(y_test, dt_y_pred)
dt_recall = recall_score(y_test, dt_y_pred)
dt_f1 = f1_score(y_test, dt_y_pred)
dt_auc = roc_auc_score(y_test, dt_y_prob)
dt_mcc = matthews_corrcoef(y_test, dt_y_pred)

# Display the evaluation metrics
print("Decision Tree Performance Metrics")
print("-" * 40)
print(f"Accuracy : {dt_accuracy:.4f}")
print(f"Precision: {dt_precision:.4f}")
print(f"Recall   : {dt_recall:.4f}")
print(f"F1 Score : {dt_f1:.4f}")
print(f"AUC Score: {dt_auc:.4f}")
print(f"MCC Score: {dt_mcc:.4f}")
```

## 2.4 K-Nearest Neighbour (KNN) Classifier

K-Nearest Neighbour (KNN) is a supervised Machine Learning algorithm used for classification and regression problems. Unlike Logistic Regression and Decision Tree, KNN does not build an explicit model during training. Instead, it classifies a new data point by identifying the 'K' nearest training samples based on distance and assigning the class that occurs most frequently among those neighbours.

In this assignment, the K-Nearest Neighbour classifier is used to predict whether a customer will subscribe to a term deposit based on the similarity of customer attributes. Since KNN is a distance-based algorithm, feature scaling is essential to ensure that all features contribute equally to the distance calculation.

```python
# =============================================================================
# 2.4.1 Train the K-Nearest Neighbour (KNN) Classifier
# =============================================================================

from sklearn.neighbors import KNeighborsClassifier

# Create the KNN classifier with K = 5
knn_model = KNeighborsClassifier(n_neighbors=5)

# Train the model using the scaled training dataset
knn_model.fit(X_train_scaled, y_train)

print("K-Nearest Neighbour model trained successfully.")
```

```python
# =============================================================================
# 2.4.2 Predict and Evaluate the K-Nearest Neighbour (KNN) Classifier
# =============================================================================

# Predict the target values for the testing dataset
knn_y_pred = knn_model.predict(X_test_scaled)

# Predict probabilities for AUC calculation
knn_y_prob = knn_model.predict_proba(X_test_scaled)[:, 1]

# Calculate evaluation metrics
knn_accuracy = accuracy_score(y_test, knn_y_pred)
knn_precision = precision_score(y_test, knn_y_pred)
knn_recall = recall_score(y_test, knn_y_pred)
knn_f1 = f1_score(y_test, knn_y_pred)
knn_auc = roc_auc_score(y_test, knn_y_prob)
knn_mcc = matthews_corrcoef(y_test, knn_y_pred)

# Display the evaluation metrics
print("K-Nearest Neighbour Performance Metrics")
print("-" * 45)
print(f"Accuracy : {knn_accuracy:.4f}")
print(f"Precision: {knn_precision:.4f}")
print(f"Recall   : {knn_recall:.4f}")
print(f"F1 Score : {knn_f1:.4f}")
print(f"AUC Score: {knn_auc:.4f}")
print(f"MCC Score: {knn_mcc:.4f}")
```

## 2.5 Naive Bayes Classifier

Naive Bayes is a supervised Machine Learning algorithm based on Bayes' Theorem. It is widely used for classification problems and predicts the class of a data instance by calculating the probability of each class given the input features. The algorithm assumes that all input features are independent of each other, which is known as the "naive" assumption.

In this assignment, the Gaussian Naive Bayes classifier is used to predict whether a customer will subscribe to a term deposit based on customer information and previous marketing campaign details. Although the independence assumption may not always hold in real-world datasets, Naive Bayes is computationally efficient and often provides good baseline classification performance.

```python
# =============================================================================
# 2.5.1 Train the Gaussian Naive Bayes Classifier
# =============================================================================

from sklearn.naive_bayes import GaussianNB

# Create the Gaussian Naive Bayes model
naive_bayes_model = GaussianNB()

# Train the model using the training dataset
naive_bayes_model.fit(X_train, y_train)

print("Gaussian Naive Bayes model trained successfully.")
```

```python
# =============================================================================
# 2.5.2 Predict and Evaluate the Gaussian Naive Bayes Classifier
# =============================================================================

# Predict the target values for the testing dataset
nb_y_pred = naive_bayes_model.predict(X_test)

# Predict probabilities for AUC calculation
nb_y_prob = naive_bayes_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
nb_accuracy = accuracy_score(y_test, nb_y_pred)
nb_precision = precision_score(y_test, nb_y_pred)
nb_recall = recall_score(y_test, nb_y_pred)
nb_f1 = f1_score(y_test, nb_y_pred)
nb_auc = roc_auc_score(y_test, nb_y_prob)
nb_mcc = matthews_corrcoef(y_test, nb_y_pred)

# Display the evaluation metrics
print("Gaussian Naive Bayes Performance Metrics")
print("-" * 45)
print(f"Accuracy : {nb_accuracy:.4f}")
print(f"Precision: {nb_precision:.4f}")
print(f"Recall   : {nb_recall:.4f}")
print(f"F1 Score : {nb_f1:.4f}")
print(f"AUC Score: {nb_auc:.4f}")
print(f"MCC Score: {nb_mcc:.4f}")
```

## 2.6 Random Forest Classifier

Random Forest is an ensemble Machine Learning algorithm that combines multiple Decision Trees to improve prediction accuracy and reduce overfitting. Instead of relying on a single Decision Tree, Random Forest builds several trees using different subsets of the training data and combines their predictions through majority voting.

In this assignment, the Random Forest classifier is used to predict whether a customer will subscribe to a term deposit based on customer information and previous marketing campaign details. Random Forest generally provides better generalization performance than a single Decision Tree and is widely used for classification problems due to its robustness and high predictive accuracy.

```python
# =============================================================================
# 2.6.1 Train the Random Forest Classifier
# =============================================================================

from sklearn.ensemble import RandomForestClassifier

random_forest_model = RandomForestClassifier(
    n_estimators=10,
    max_depth=8,
    random_state=42
)

# Train the model using the training dataset
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully.")
```

```python
# =============================================================================
# 2.6.2 Predict and Evaluate the Random Forest Classifier
# =============================================================================

# Predict the target values for the testing dataset
rf_y_pred = random_forest_model.predict(X_test)

# Predict probabilities for AUC calculation
rf_y_prob = random_forest_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
rf_accuracy = accuracy_score(y_test, rf_y_pred)
rf_precision = precision_score(y_test, rf_y_pred)
rf_recall = recall_score(y_test, rf_y_pred)
rf_f1 = f1_score(y_test, rf_y_pred)
rf_auc = roc_auc_score(y_test, rf_y_prob)
rf_mcc = matthews_corrcoef(y_test, rf_y_pred)

# Display the evaluation metrics
print("Random Forest Performance Metrics")
print("-" * 45)
print(f"Accuracy : {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall   : {rf_recall:.4f}")
print(f"F1 Score : {rf_f1:.4f}")
print(f"AUC Score: {rf_auc:.4f}")
print(f"MCC Score: {rf_mcc:.4f}")
```

```python
# =============================================================================
# Step 4 - Model Evaluation
# =============================================================================

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree_model,
    "KNN": knn_model,
    "Naive Bayes": naive_bayes_model,
    "Random Forest": random_forest_model
}

evaluation_results = []

for model_name, model in models.items():

    y_pred = model.predict(X_test)

    # AUC requires probabilities if available
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
        auc = roc_auc_score(y_test, y_prob)
    else:
        auc = roc_auc_score(y_test, y_pred)

    evaluation_results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "AUC Score": auc,
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred),
        "MCC Score": matthews_corrcoef(y_test, y_pred)
    })

evaluation_df = pd.DataFrame(evaluation_results)

evaluation_df
```
